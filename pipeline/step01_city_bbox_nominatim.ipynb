Fetch the bounding box of each city in the city list (Skyscraper Center ranking) from the OpenStreetMap Nominatim API and save the list with the boxes.

```python
import pandas as pd
import requests
import time
```

```python
def get_bounding_box(city_name, country):
    url = f"https://nominatim.openstreetmap.org/search?city={city_name}&country={country}&format=json&limit=1"
    headers = {'User-Agent': 'City BBox Fetcher'}  # Set a custom User-Agent
    response = requests.get(url, headers=headers)
    if response.status_code == 200:
        data = response.json()
        if data:
            bbox = data[0]['boundingbox']
            return {
                'min_lat': bbox[0],
                'max_lat': bbox[1],
                'min_lon': bbox[2],
                'max_lon': bbox[3]
            }
    return None
```

```python
bbox_Berlin= get_bounding_box("Berlin","Germany")
```

```python
bbox_Berlin
```

```python
# Read the Excel file
file_path = '<CITY_LIST_DIR>/cities_list.xlsx'
cities_data = pd.read_excel(file_path, skiprows=2)
cities_data.columns = [
    'Rank', 'City', 'Country', 'Population', 'Area', 'Density',
    'Buildings_150m+', 'Buildings_200m+', 'Buildings_300m+',
    'Min_Longitude', 'Min_Latitude', 'Max_Longitude', 'Max_Latitude'
]
```

```python
cities_data
```

```python
# Fetch bounding boxes and update the dataframe
for index, row in cities_data.iterrows():
    bbox = get_bounding_box(row['City'], row['Country'])
    if bbox:
        cities_data.at[index, 'Min_Latitude'] = bbox['min_lat']
        cities_data.at[index, 'Max_Latitude'] = bbox['max_lat']
        cities_data.at[index, 'Min_Longitude'] = bbox['min_lon']
        cities_data.at[index, 'Max_Longitude'] = bbox['max_lon']
    time.sleep(1)  # Respect the API rate limit
```

```python
cities_data
```

```python
# Save the updated dataframe to a new Excel file
output_file_path = '<CITY_LIST_DIR>/cities_list_bbox.xlsx'
cities_data.to_excel(output_file_path, index=False)
```

```python
def get_bounding_box(city_name):
    # Construct the URL for the API request
    url = f"https://nominatim.openstreetmap.org/search?city={city_name}&format=json&polygon_geojson=1&limit=1"
    # Send a GET request to the Nominatim API
    response = requests.get(url)
    
    if response.status_code != 200:
        return f"Error: Received status code {response.status_code}"
    
    try:
        data = response.json()  # Parse the JSON response
    except ValueError as e:
        print(f"Raw response content: {response.text}")
        return f"Error decoding JSON: {e}"
    
    # Check if data is received
    if data:
        # Extract the bounding box
        bbox = data[0]['boundingbox']
        return {
            'south_latitude': bbox[0],
            'north_latitude': bbox[1],
            'west_longitude': bbox[2],
            'east_longitude': bbox[3]
        }
    else:
        return "No data found for the specified city."

bbox_Berlin = get_bounding_box("Berlin")
print(bbox_Berlin)
```
